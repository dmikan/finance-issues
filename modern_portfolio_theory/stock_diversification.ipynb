```python
import numpy as np
import pandas as pd
import yfinance as yf
import plotly.graph_objects as go
```

```python
# Definir lista de activos del S&P 500
stockList = [
    'AAPL', 'MSFT', 'GOOGL', 'AMZN', 'NVDA', 'META', 'BRK-B', 'JNJ', 'JPM', 'V',
    'PG', 'UNH', 'HD', 'MA', 'BAC', 'DIS', 'PFE', 'XOM', 'KO', 'PEP',
    'CSCO', 'WMT', 'COST', 'MCD', 'CRM', 'TMO', 'ABT', 'ADBE', 'ACN', 'NKE'
]
market_ticker = '^GSPC'
```

```python
# Descargar datos históricos (últimos 5 años)
data = yf.download(stockList, 
                   period='5y',
                   auto_adjust=True,     
                   multi_level_index=False)['Close']
```

```python
data.tail(5)
```

```python
# Retornos diarios de los activos
returns = data.pct_change().dropna()
returns.tail(5)
```

```python
# Matriz de covarianza diaria y métricas base
cov_matrix = returns.cov()
cov_matrix.head(5)
```

```python
variances = np.diag(cov_matrix)
print(variances.round(6))
```

```python
# Varianza promedio individual (sigma^2 bar) y covarianza promedio (sigma_ij bar)
N_total = len(stockList)
var_avg = np.mean(variances)
cov_avg = (cov_matrix.sum().sum() - np.sum(variances)) / (N_total * (N_total - 1))
print(f"Varianza promedio individual: {var_avg:.6f}")
print(f"Covarianza promedio: {cov_avg:.6f}")
```

```python
# Convertir a volatilidad anualizada en porcentaje (%)
trading_days = 252
n_range = np.arange(1, N_total + 1)

# Formula analítica de varianza para un portafolio de pesos iguales (1/n)
var_portfolio = (1 / n_range) * var_avg + ((n_range - 1) / n_range) * cov_avg
vol_portfolio_annual = np.sqrt(var_portfolio * trading_days) * 100
```

```python
# Límite asíntotico: Riesgo Sistemático (No diversificable)
systematic_risk = np.sqrt(cov_avg * trading_days) * 100
```

```python
# 3. Generar gráfico interactivo en Plotly
fig = go.Figure()

# Sombreado de Riesgo No Diversificable (Base asintótica)
fig.add_trace(go.Scatter(
    x=n_range,
    y=[systematic_risk] * N_total,
    mode='lines',
    name=f'Riesgo Sistemático (~{systematic_risk:.1f}%)',
    line=dict(color='#d62728', dash='dash', width=2),
    fill='tozeroy',
    fillcolor='rgba(103, 169, 207, 0.25)',
    hoverinfo='skip'
))

# Curva de Riesgo Total con sombreado de Riesgo Diversificable
fig.add_trace(go.Scatter(
    x=n_range,
    y=vol_portfolio_annual,
    mode='lines+markers',
    name='Riesgo Total del Portafolio',
    line=dict(color='#1f77b4', width=3),
    marker=dict(size=6, color='#1f77b4'),
    fill='tonexty',
    fillcolor='rgba(239, 138, 98, 0.4)',
    hovertemplate='<b>Activos (n):</b> %{x}<br><b>Volatilidad Total:</b> %{y:.2f}%<extra></extra>'
))

# Ajustes de layout visual
fig.update_layout(
    template='plotly_dark',
    title=dict(
        text='<b>EFECTO DE LA DIVERSIFICACIÓN EN EL RIESGO</b>',
        font=dict(size=16, color='#ffffff')
    ),
    xaxis=dict(
        title='Número de Activos (n)',
        dtick=2,
        gridcolor='#2a2d3e',
        zerolinecolor='#2a2d3e'
    ),
    yaxis=dict(
        title='Desviación Estándar Anual Promedio (%)',
        gridcolor='#2a2d3e',
        zerolinecolor='#2a2d3e'
    ),
    hovermode='x unified',
    paper_bgcolor='#111319',
    plot_bgcolor='#111319',
    legend=dict(
        x=0.55,
        y=0.95,
        bgcolor='rgba(22, 27, 46, 0.85)',
        bordercolor='#3a3f58',
        borderwidth=1,
        font=dict(color='#e0e0e0')
    ),
    width=900,
    height=520
)

fig.show()
```
