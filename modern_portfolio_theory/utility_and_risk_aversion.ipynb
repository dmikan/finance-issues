```python
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
# 1. Definición de parámetros
sigma_neg = np.linspace(-15, 0, 50)  # Lado negativo (matemático)
sigma_pos = np.linspace(0, 15, 50)   # Lado positivo (financiero real)
E_0_levels = [2, 3, 4, 5, 6]
gamma = 50.0

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']
```

```python
# ==========================================
# 2. Figura 1: Mapa de Curvas de Indiferencia (2D)
# ==========================================
fig_2d = go.Figure()

for idx, e0 in enumerate(E_0_levels):
    # Lado negativo (Gris / Punteado)
    E_neg = e0 + 0.5 * gamma * (sigma_neg**2) / 100
    fig_2d.add_trace(go.Scatter(
        x=sigma_neg, y=E_neg,
        mode='lines',
        line=dict(width=2, color='gray', dash='dot'),
        showlegend=False,
        hoverinfo='skip'
    ))

    # Lado positivo (Color / Sólido)
    E_pos = e0 + 0.5 * gamma * (sigma_pos**2) / 100
    fig_2d.add_trace(go.Scatter(
        x=sigma_pos, y=E_pos,
        mode='lines',
        name=f'U = {e0}%',
        line=dict(width=3, color=colors[idx])
    ))

# Línea vertical de frontera en sigma = 0
fig_2d.add_vline(x=0, line_width=1.5, line_dash="dash", line_color="black")

fig_2d.update_layout(
    title=f'Curvas de Indiferencia 2D (Gris: σ < 0 | Color: σ ≥ 0)',
    xaxis_title='Desviación Estándar σ (%)',
    yaxis_title='Retorno Esperado E (%)',
    yaxis=dict(range=[0, 10]),
    xaxis=dict(range=[-15, 15]),
    template='plotly_white'
)
fig_2d.show()
```

```python
# ==========================================
# 3. Figura 2: Superficie de Utilidad 3D U(E, σ)
# ==========================================
# Malla lado negativo (σ < 0)
sigma_grid_neg, E_grid_neg = np.meshgrid(np.linspace(-15, 0, 30), np.linspace(0, 10, 30))
U_grid_neg = E_grid_neg - 0.5 * gamma * (sigma_grid_neg**2) / 100

# Malla lado positivo (σ ≥ 0)
sigma_grid_pos, E_grid_pos = np.meshgrid(np.linspace(0, 15, 30), np.linspace(0, 10, 30))
U_grid_pos = E_grid_pos - 0.5 * gamma * (sigma_grid_pos**2) / 100

fig_3d = go.Figure()

# Superficie en escala de grises (Lado no físico)
fig_3d.add_trace(go.Surface(
    x=sigma_grid_neg, 
    y=E_grid_neg, 
    z=U_grid_neg,
    colorscale='Greys',
    showscale=False,
    opacity=0.6,
    name='σ < 0 (Matemático)'
))

# Superficie a color Viridis (Lado positivo relevante)
fig_3d.add_trace(go.Surface(
    x=sigma_grid_pos, 
    y=E_grid_pos, 
    z=U_grid_pos,
    colorscale='Viridis',
    showscale=True,
    colorbar=dict(title='U'),
    name='σ ≥ 0 (Relevante)'
))

fig_3d.update_layout(
    title='Superficie de Utilidad 3D U(E, σ)',
    scene=dict(
        xaxis_title='σ (%)',
        yaxis_title='E (%)',
        zaxis_title='Utilidad U'
    ),
    template='plotly_white'
)

# Mostrar gráficos
fig_3d.show()
```
